```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("heart_failure_clinical_records_dataset.csv")

df["ejection_fraction"].plot.kde()

plt.xlabel("Ejection fraction")
plt.ylabel("Density")
plt.title("Kernel Density Estimate of Ejection Fraction")
plt.show()
```

```python
df["ejection_fraction"].plot.kde()

plt.xlabel("Ejection fraction")
plt.ylabel("Density")
plt.title("Kernel Density Estimate of Ejection Fraction")
plt.show()
```

```python
df.sample(frac=1.0, replace=True)

import matplotlib.pyplot as plt

for i in range(15):
    sample = df.sample(frac=1.0, replace=True)
    sample["ejection_fraction"].plot.kde()

plt.xlabel("Ejection fraction")
plt.ylabel("Density")
plt.title("15 Bootstrap KDEs")
plt.show()
```

Our original KDE is more reliable when the resampled KDE curves are close togehter. We can notice this around 30 to 45 since the curves from the 15 resampled datasets are closer together in this range. However, from 70 to 80, we notice that there is more variation in the curves because we very few observations with this range. So, the KDE is less stable there.
